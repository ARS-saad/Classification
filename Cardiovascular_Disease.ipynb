```python
x = df.iloc[:, 1:-1]
x
```

```python
y = df[["cardio"]]
y
```

# Scalling data set

```python
from sklearn.preprocessing import StandardScaler
```

```python
feature_x = StandardScaler().fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split, KFold, cross_val_score
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=.3, random_state=52)
```

```python
xtrain
```

# Train model

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rclf = RandomForestClassifier()
```

```python
rclf.fit(xtrain, ytrain)
```

```python
rclf.score(xtest, ytest)
```

```python
kf = KFold(n_splits=5, random_state=51, shuffle=True)
```

```python
kf_score = cross_val_score(rclf, x, y, cv=kf)
kf_score
```

```python
kf_score.mean()
```

# Save Machine Learning Models

```python
import pickle as pk

with open('My_Model1','wb') as file:
    pk.dump(rclf, file)

# pk.dump(rclf, open('My_Model1', 'wb'))
```
